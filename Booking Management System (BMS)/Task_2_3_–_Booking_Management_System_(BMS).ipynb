```python
# Task 2.3 – Booking Management System (BMS)

new_car_id = [
    ["13-04-2025", "الأحد", [6, 10, 1, 10, 7]],
    ["14-04-2025", "الاثنين", [10, 2, 10, 6, 4]],
    ["15-04-2025", "الثلاثاء", [3, 1, 10, 10, 9]],
    ["16-04-2025", "الأربعاء", [10, 8, 10, 7, 3]],
    ["17-04-2025", "الخميس", [2, 10, 4, 10, 1]]
]

# أوقات المواعيد
times = ["9:00", "10:00", "11:00", "12:00", "1:00"]

print("\n--- نظام حجز تراخيص السيارات ---")

# 1. عرض الأيام المتاحة
print("\nالأيام المتاحة للحجز:")
for i, day in enumerate(new_car_id, 1):
    print(f"{i}. {day[1]} ({day[0]})")

# 2. اختيار اليوم
while True:
    try:
        day_choice = int(input("اختر رقم اليوم: "))
        if 1 <= day_choice <= len(new_car_id):
            selected_day = new_car_id[day_choice - 1]
            break
        else:
            print("الرجاء اختيار رقم من القائمة")
    except:
        print("الرجاء إدخال رقم صحيح")

# 3. عرض المواعيد المتاحة
print(f"\nالمواعيد المتاحة ليوم {selected_day[1]}:")
available_slots = []

for i, (time, booked) in enumerate(zip(times, selected_day[2])):
    available = 10 - booked
    if available > 0:
        print(f"{i+1}. {time} (الأماكن المتبقية: {available})")
        available_slots.append(i)

if not available_slots:
    print("لا توجد مواعيد متاحة في هذا اليوم")
    exit()

# 4. اختيار الموعد
while True:
    try:
        slot_choice = int(input("\nاختر رقم الموعد: "))
        if 1 <= slot_choice <= len(times) and (slot_choice - 1) in available_slots:
            selected_time = times[slot_choice - 1]
            break
        else:
            print("الرجاء اختيار رقم موعد متاح")
    except:
        print("الرجاء إدخال رقم صحيح")

# 5. إدخال البيانات الشخصية
print("\n--- بيانات الحجز ---")
full_name = input("الاسم بالكامل: ")
id_number = input("رقم الهوية: ")

# 6. طباعة تأكيد الحجز
print("\n--- تأكيد الحجز ---")
print(f"اليوم: {selected_day[1]} ({selected_day[0]})")
print(f"الموعد: {selected_time}")
print(f"الاسم: {full_name}")
print(f"رقم الهوية: {id_number}")
print("\nتم الحجز بنجاح! شكراً لاستخدامكم نظامنا")
```
