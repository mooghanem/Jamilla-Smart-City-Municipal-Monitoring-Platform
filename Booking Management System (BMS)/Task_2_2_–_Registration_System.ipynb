```python
# Task 2.2 – Registration System

users = {}

print("--- نظام التسجيل ---")

# 1. Username
username = input("ادخل اسم المستخدم: ")

# 2. Password
while True:
    password = input("ادخل كلمة المرور (يجب أن تحتوي على حروف كبيرة وصغيرة): ")


    has_upper = False
    has_lower = False

    for char in password:
        if char.isupper():
            has_upper = True
        elif char.islower():
            has_lower = True

    if has_upper and has_lower:
        break
    else:
        print("كلمة المرور يجب أن تحتوي على حروف كبيرة وصغيرة")

while True:
    confirm = input("أعد إدخال كلمة المرور: ")
    if password == confirm:
        users[username] = password
        print("تم التسجيل بنجاح!")
        break
    else:
        print("كلمة المرور غير متطابقة، حاول مرة أخرى")
```
